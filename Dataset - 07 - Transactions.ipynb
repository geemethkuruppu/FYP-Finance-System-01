## Transactions

```python
import pandas as pd

df_8 = pd.read_csv("../Datasets/transactions.csv")
df_8.head()
```

```python
df_8.info()
```

```python
# Count number of transactions per customer
customer_counts = df_8["CUST_ID"].value_counts()

# Filter customers with more than 1000 transactions
customers_over_1000 = customer_counts[customer_counts > 1000]

print("Customers with over 1000 transactions:")
print(customers_over_1000)
```

```python
# total transaction for all customers 
customers_over_1000.sum()
```

```python
# Create the new dataset containing ONLY those customers
df_8 = df_8[df_8["CUST_ID"].isin(customers_over_1000.index)]
```

```python
df_8.info()
```

```python
df_8.head()
```

```python
# Create TransactionID for each transaction
df_8 = df_8.reset_index(drop=True)

df_8["TransactionID"] = df_8.index + 1
df_8["TransactionID"] = df_8["TransactionID"].apply(lambda x: f"TRX8_{str(x).zfill(6)}")
```

```python
df_8.head()
```

```python
# Create a mapping from CUST_ID → US8001, US8002, ...
df_8 = df_8.reset_index(drop=True)

# Get unique customers
unique_customers = df_8["CUST_ID"].unique()

# Create mapping: each customer → US8001, US8002, ...
user_map = {cust: f"US8{str(i+1).zfill(3)}" for i, cust in enumerate(unique_customers)}

user_map
```

```python
df_8["UserID"] = df_8["CUST_ID"].map(user_map)
df_8 = df_8.drop(columns=["CUST_ID"])

cols = ["UserID"] + [c for c in df_8.columns if c != "UserID"]
df_8 = df_8[cols]

df_8.head()
```

```python
# check all EXP_TYPE values
df_8["EXP_TYPE"].unique()
```

```python
# Rename EXP_TYPE to Category
df_8 = df_8.rename(columns={"EXP_TYPE": "Category"})
df_8.head()
```

```python
# Create Type column
expense_categories = [
    'Motor/Travel', 'Housing', 'Entertainment', 'Groceries', 'Savings',
    'Bills and Utilities', 'Clothing', 'Tax', 'Health', 'Fines',
    'Education', 'Gambling'
]

df_8["Type"] = df_8["Category"].apply(
    lambda x: "Expense" if x in expense_categories else "Income"
)
```

```python
df_8[["Category", "Type"]].head(20)
```

```python
df_8.head()
```

```python
# creating Merchant column 
df_8["Merchant"] = " "
df_8.head()
```

```python
# creating Account Name column 
df_8["Account_Name"] = " "
df_8.head()
```

```python
# Generate Descriptions Randomly

import numpy as np

description_map = {
    "Motor/Travel": [
        "Taxi fare to destination",
        "Fuel purchase at local station",
        "Bus fare payment",
        "Ride service payment",
        "Parking fee"
    ],
    "Housing": [
        "Monthly rent payment",
        "House maintenance charges",
        "Property service fee"
    ],
    "Entertainment": [
        "Movie ticket purchase",
        "Online streaming subscription",
        "Concert or event payment"
    ],
    "Groceries": [
        "Grocery shopping at supermarket",
        "Bought vegetables and fruits",
        "Household daily essentials"
    ],
    "Savings": [
        "Transfer to savings account",
        "Automatic savings deposit"
    ],
    "Bills and Utilities": [
        "Electricity bill payment",
        "Water bill settlement",
        "Internet bill payment",
        "Gas bill charges"
    ],
    "Clothing": [
        "Clothing purchase",
        "Bought new apparel",
        "Fashion store transaction"
    ],
    "Tax": [
        "Income tax payment",
        "Government tax deduction",
        "Local tax settlement"
    ],
    "Health": [
        "Pharmacy medical supplies",
        "Hospital consultation payment",
        "Lab test fees"
    ],
    "Fines": [
        "Traffic fine payment",
        "Late fee charge",
        "Government penalty fee"
    ],
    "Education": [
        "Course enrollment fee",
        "Online class subscription",
        "School-related payment"
    ],
    "Gambling": [
        "Casino game transaction",
        "Online betting payment",
        "Lottery ticket purchase"
    ]
}

def generate_description(category):
    if category in description_map:
        return np.random.choice(description_map[category])
    return "Miscellaneous transaction"
    

df_8["Transaction Description"] = df_8["Category"].apply(generate_description)

df_8[["Category", "Transaction Description"]].head(20)
```

```python
df_8.head()
```

```python
# creating Currency  column 
df_8["Currency "] = " "
df_8.head()
```

```python
# Rename DATE and AMOUNT 
df_8 = df_8.rename(columns={"DATE": "Date"})
df_8 = df_8.rename(columns={"AMOUNT": "Amount"})
df_8 = df_8.rename(columns={"Account_Name": "Account Name"})
df_8.head()
```

```python
# Check for NULL values in each required column
df_8[["Date", "Category", "Amount", "UserID", "TransactionID"]].isna().sum()
```

```python
# Check for EMPTY string values (“ ”) in those columns
(df_8[["Date", "Category", "Amount", "UserID", "TransactionID"]] == "").sum()
```

```python
# Check for rows where columns are BLANK or NULL

missing_rows = df_8[
    df_8["Date"].isna() |
    df_8["Category"].isna() |
    df_8["Amount"].isna() |
    df_8["UserID"].isna() |
    df_8["TransactionID"].isna() |
    (df_8["Category"].str.strip() == "") |
    (df_8["UserID"].str.strip() == "")
]

missing_rows.head()
```

```python
# Check if Amount has invalid values (0, negative, NaN

df_8["Amount"] = pd.to_numeric(df_8["Amount"], errors="coerce")

# Check invalid values
invalid_amounts = df_8[
    df_8["Amount"].isna() | (df_8["Amount"] <= 0)
]

print("Invalid Amount Rows:", len(invalid_amounts))
invalid_amounts.head()
```

```python
print("Null check:")
print(df_8.isna().sum())

print("\nEmpty string check:")
print((df_8 == "").sum())

print("\nTotal rows:", len(df_8))
print("Rows with ANY missing critical fields:", len(missing_rows))
```
